```python
import os
os.environ["TOKENIZERS_PARALLELISM"] = "true"
from unstructured.partition.auto import partition

# when the file is a directory
filename = "data/ELIGIBLE_RFP_2.pdf"
elements = partition(filename=filename, content_type="application/pdf")
print("".join(str(e) for e in elements))
    
# # when the file is a file object
# with open(f"./data/world-war-one.pdf","rb") as fobj:
#     elements = partition(file=fobj, include_page_breaks=True)
    
# #print the extracted elements
# for e in elements:
#     print(e)
```

```python
#print the extracted elements
from unstructured.cleaners.core import clean
from unstructured.cleaners.core import clean_non_ascii_chars
from unstructured.cleaners.core import remove_punctuation
from unstructured.cleaners.core import replace_unicode_quotes
from unstructured.cleaners.core import group_broken_paragraphs

for e in elements:
    if hasattr(e, 'text') and isinstance(e.text, str):  # Ensure the element has a text attribute and it's a string
        text = e.text
        text = clean(text, bullets=True, lowercase=True, extra_whitespace=True, dashes=True)
        text = clean_non_ascii_chars(text)
        text = remove_punctuation(text)
        group_broken_paragraphs(text)  # Uncomment if needed
        text = replace_unicode_quotes(text)
        print(text)
    else:
        print(f"Skipping element of type {type(e)} as it does not have a text attribute.")
```

```python
from unstructured.partition.auto import partition
import json

# 🔁 Load any unstructured document (PDF, DOCX, etc.)
elements = partition(filename="data/ELIGIBLE_RFP_1.pdf")  # or .docx
structured_output = []

for element in elements:
    el_dict = {
        "type": element.category,
        "text": element.text,
        "metadata": element.metadata.to_dict()
    }

    # Special handling for tables
    if element.category == "Table":
        el_dict["html"] = element.metadata.text_as_html

    structured_output.append(el_dict)

# Save the output
with open("rfp_structured_output_with_tables.json", "w", encoding="utf-8") as f:
    json.dump(structured_output, f, indent=2)
```
